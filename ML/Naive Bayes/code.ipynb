```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
df.drop(['Name','Ticket','Cabin','Embarked','PassengerId','SibSp','Parch'] , axis='columns' , inplace=True)
df
```

```python
inputs = df.drop(['Survived'] , axis=1)
```

```python
target = df.Survived
```

```python
dummies = pd.get_dummies(inputs.Sex , dtype=int)
# dtype int gives you 0/1 , or else it will be true or false that is boolean
dummies
```

```python
inputs = pd.concat([inputs , dummies] , axis=1)
```

```python
inputs.drop(['Sex'] , axis=1 , inplace=True)
```

```python
inputs
```

```python
inputs.isnull().sum()
```

```python
inputs.Age.fillna(inputs.Age.mean() , inplace=True)
```

```python
inputs.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split

x_train , x_test , y_train ,y_test = train_test_split(inputs , target , test_size = 0.2)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()

model.fit(x_train , y_train)

model.score(x_test , y_test)
```
