```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn import linear_model
```

```python
df = pd.read_csv('homeprices.csv')
df
```

```python
median = df['bedrooms'].median()
```

```python
df.fillna({'bedrooms': median} , inplace=True)
```

```python
df
```

```python
reg = linear_model.LinearRegression()
```

```python
reg.fit(df[['area' , 'bedrooms' , 'age']] , df.price)
# we will predict price 
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
# we want 3000 sqr ft area , 3 bedrooms , 40 year old
reg.predict([[3000 , 3 ,40]])
```

```python
# price  = m1 * area + m2 * bedrooms + m3 * age + b
```

```python
import pandas as pd
new_df = pd.read_csv('hiring.csv')
new_df
```

```python
new_df['experience'] = new_df['experience'].fillna('zero')
```

```python
new_df
```

```python
new_df['test_score(out of 10)'].fillna(new_df['test_score(out of 10)'].mean() , inplace=True)
```

```python

new_df
```

```python
from word2number import w2n
new_df['experience_words'] = new_df['experience'].apply(w2n.word_to_num)
```

```python
new_df
```

```python
new_df.drop(columns=['experience'] ,axis=1)
```

```python
from sklearn import linear_model
model = linear_model.LinearRegression()
```

```python
model.fit(new_df[['experience_words' , 'test_score(out of 10)',	'interview_score(out of 10)']] , new_df['salary($)'])
```

```python
model.predict([[2,9,6]])
```
