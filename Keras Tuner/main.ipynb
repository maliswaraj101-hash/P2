```python
import pandas as pd
```

```python

df =  pd.read_csv('diabetes.csv')
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.iloc[: , :-1].values
y = df.iloc[: , -1].values
```

```python
y
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x = scaler.fit_transform(x)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=40)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(32 , activation='relu' , input_shape=(8 , )))
model.add(Dense(1 , activation='sigmoid'))

model.compile(optimizer='Adam' , loss = 'binary_crossentropy' , metrics=['accuracy'])
```

```python
model.fit(x_train , y_train , batch_size= 32 , epochs=10 )
```

```python
import kerastuner as kt
```

```python
# hp means hyperparameter
def build_model(hp):
    
    model = Sequential()
    model.add(Dense(32 , activation='relu' , input_shape = (8 ,)))
    model.add(Dense(1 , activation='sigmoid'))

    optimizer = hp.Choice('optimizer' , values = ['adam' , 'sgd' , 'rmsprop' , 'adadelta '])
    # optimiser tries all comvinations from these values

    model.compile(optimizer = optimizer , loss='binary_crossentropy' , metrics=['accuracy'] )

    return model

    
```

```python
tuner = kt.RandomSearch(build_model , objective='val_accuracy' , max_trials=5 , directory = 'my_dir' , project_name = 'intro_to_kt' , overwrite = True)
```

```python
 
tuner.search(x_train , y_train , epochs = 5 , validation_data = (x_test , y_test) , verbose = 1)
# 5 trials and 5 epochs means each trial will contains 5 epochs
```

```python
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]
print(best_hps.values)
```

```python
tuner.results_summary()
```

```python
all_trials = tuner.oracle.trials
```

```python
for trial_id , trial in all_trials.items():
    print(f"Trial id : {trial_id}")
    print(f"Hyperparameters : {trial.hyperparameters.values}")
    print(f"score : {trial.score}")
```

```python
def build_model(hp):
    model = Sequential()

    units = hp.Int('units' , min_value = 8 ,max_value =  128 , step = 8)
    # first 8 neurons then 16 then 24 then 32....till 128

    model.add(Dense(units=units , activation='relu' , input_shape = (8,)))
    # hp. int means it will check for range between 8 and 128 with step size 8 , 8 - 16 - 24 - 32 - 40 till 128
    model.add(Dense(1 , activation='sigmoid'))

    model.compile(optimizer='rmsprop' , loss='binary_crossentropy' , metrics=['accuracy'])

    return model
```

```python
tuner = kt.RandomSearch(build_model , objective='val_accuracy' , max_trials=5)
```

```python
tuner.search(x_train , y_train , epochs = 5 , validation_data = (x_test , y_test))
```

```python
all_trials = tuner.oracle.trials
```

```python
for trial_id , trial in all_trials.items():
    print(f"Trial id : {trial_id}")
    print(f"Hyperparameters : {trial.hyperparameters.values}")
    print(f"score : {trial.score}")
```

```python
def build_model_num(hp):

    model = Sequential()

    model.add(Dense(120 , activation='relu' , input_shape = (8,)))

    for i in  range(hp.Int('num_layers' , min_value = 1 , max_value = 10 )) :

        model.add(Dense(120 , activation='relu'))
    
    model.add(Dense(1 , activation='sigmoid'))

    model.compile(optimizer='rmsprop' , loss='binary_crossentropy' , metrics=['accuracy'])

    return model
```

```python
tuner_num = kt.RandomSearch(build_model_num , objective='val_accuracy' , max_trials=3 )
```

```python
tuner_num.search(x_train , y_train , epochs = 5 , validation_data = (x_test , y_test))
```

```python
def build_model(hp):

    model = Sequential()
    
    counter = 0

    for i in range(hp.Int('num_layers' , min_value = 1 , max_value = 10)):
        if counter == 0 :

            model.add(Dense(hp.Int('units' + str(i) , min_value = 8 , max_value = 128 , step = 8) , activation=hp.Choice('activation' + str(i) , values = ['relu' , 'tanh' , 'sigmoid']) , input_shape = (8 , )))
            # model.add(Dropout(hp.Choice('dropout' +  str(i) , values = [0.1,0.2,0.3])))
        else :

            model.add(Dense(hp.Int('units' + str(i) , min_value = 8 , max_value = 128 , step = 8) , activation=hp.Choice('activation' + str(i) , values = ['relu' , 'tanh' , 'sigmoid'])))
            # model.add(Dropout(hp.Choice('dropout' +  str(i) , values = [0.1,0.2,0.3])))

        counter = counter + 1
    
    model.add(Dense(1 , activation='sigmoid'))

    model.compile(optimizer=hp.Choice('optimizer' , values = ['rmsprop' , 'adam' , 'sgd' , 'nadam']) , loss='binary_crossentropy' , metrics=['accuracy'])

    return model
```

```python
tuner = kt.RandomSearch(build_model , objective='val_accuracy' , max_trials= 3)
```

```python
tuner.search(x_train , y_train , epochs = 5 , validation_data = (x_test , y_test))
```
