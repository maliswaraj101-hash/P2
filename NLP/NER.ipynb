```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
# this is a pretrained model
```

```python
nlp.pipe_names
```

```python
doc = nlp("Tesla Inc is going to acquire Twitter Inc for 45 billion dollar ")
```

```python
doc.ents
```

```python
for ent in doc.ents:
    print(ent.text , "|" , ent.label_ , "|" , spacy.explain(ent.label_))
```

```python
from spacy import displacy

displacy.render(doc , style='ent')
```

```python
doc[0:5]
```

```python
from spacy.tokens import Span
```

```python
s1 = Span(doc , 0 , 1 , label='ORG')
# 0 to 1 means Tesla in our doc
s2 = Span(doc , 6 , 7 , label="ORG")
# 6 to 7 is twitter

doc.set_ents([s1 , s2] , default='unmodified')
# we are setting words are custom entities as we want
```

```python
for ent in doc.ents:
    print(ent.text , "|" , ent.label_ , "|" , spacy.explain(ent.label_))
    # earlier it was recognising twitter as person now as ORG
```
