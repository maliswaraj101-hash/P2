```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(u"I will google about facebook")
# this already does the POS Tagging
```

```python
doc.text
```

```python
doc[0]
```

```python
doc[0].pos_
# pronoun
```

```python
doc[-1]
```

```python
spacy.explain('PRON')
```

```python
for word in doc:
    print(word.text , "------->" , word.pos_ , word.tag_ , spacy.explain(word.tag_))
```
